```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Load data
path = 'data/processed/train.csv'
try:
    df = pd.read_csv(path)
    print('Loaded', path)
except FileNotFoundError:
    df = pd.DataFrame()
    print('File not found — replace `path` with your data file.')

df.head()
```

```python
# Quick overview
print('Shape:', df.shape)
print('\nMissing values by column:')
print(df.isna().sum())

# Example churn distribution (assumes `churn` column)
if 'churn' in df.columns:
    display(df['churn'].value_counts(normalize=True))
    sns.countplot(data=df, x='churn')
    plt.title('Churn distribution')
    plt.show()

# Add more EDA steps below
```

```python
# Feature distributions
# Plot histograms for numeric features; use `churn` as hue if present
num_cols = df.select_dtypes(include=['number']).columns.tolist()
if 'churn' in num_cols:
    num_cols = [c for c in num_cols if c != 'churn']

if len(num_cols) == 0:
    print('No numeric columns to plot')
else:
    for col in num_cols:
        plt.figure(figsize=(6,3))
        if 'churn' in df.columns:
            sns.histplot(data=df, x=col, hue='churn', kde=True, stat='density', common_norm=False)
        else:
            sns.histplot(data=df, x=col, kde=True)
        plt.title(f'Distribution of {col}')
        plt.tight_layout()
        plt.show()
```

```python
# Correlation heatmap and top correlations with churn
if df.shape[0] == 0:
    print('No data loaded')
else:
    corr = df.select_dtypes(include=['number']).corr()
    if corr.empty:
        print('No numeric correlations available')
    else:
        plt.figure(figsize=(10,8))
        sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
        plt.title('Feature correlation matrix')
        plt.show()

# If churn in columns, show top correlations with churn
if 'churn' in df.columns:
    churn_corr = corr['churn'].drop('churn', errors='ignore').abs().sort_values(ascending=False)
    display(churn_corr.head(10))
```
